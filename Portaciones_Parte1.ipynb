```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd

# Ruta corregida basada en la ubicación real de tus archivos
ruta_base = '/content/drive/MyDrive/TrabajoGrado/'

# Carga de archivos especificando el separador correcto ';'
df_donante = pd.read_csv(ruta_base + 'Operaciones_Portación_Donante.csv', sep=';')
df_receptor = pd.read_csv(ruta_base + 'Operaciones_Portación_Receptor.csv', sep=';')
df_rechazos = pd.read_csv(ruta_base + 'OPERACIONES_RECHAZADAS_PNM.csv', sep=';')
df_neto = pd.read_csv(ruta_base + 'PNM_RECEPTOR_NETO.csv', sep=';')
```

## Duplicados

```python
duplicados = df_donante.duplicated().sum()
print(f"\n**Filas Duplicadas donantes:** {duplicados} ({round((duplicados / len(df_donante)) * 100, 2)}% del total)")
duplicados = df_receptor.duplicated().sum()
print(f"\n**Filas Duplicadas receptor:** {duplicados} ({round((duplicados / len(df_receptor)) * 100, 2)}% del total)")
duplicados = df_rechazos.duplicated().sum()
print(f"\n**Filas Duplicadas rechazos:** {duplicados} ({round((duplicados / len(df_rechazos)) * 100, 2)}% del total)")
duplicados = df_neto.duplicated().sum()
print(f"\n**Filas Duplicadas neto:** {duplicados} ({round((duplicados / len(df_neto)) * 100, 2)}% del total)")
```

```python
filas_duplicadas = df_rechazos[df_rechazos.duplicated(keep='first')] #False para mostrar duplicados como original. 'first' para mostrar solo adicionales.
display(filas_duplicadas)
```

## Estandarización nombres de empresa

```python
validacion = df_donante.groupby('ID_EMPRESA')['EMPRESA'].nunique().reset_index()
inconsistencias = validacion[validacion['EMPRESA'] > 1]
if len(inconsistencias) > 0:
    print(f"Se encontraron {len(inconsistencias)} NITs con nombres de empresa diferentes:")

    # Filtramos el DataFrame original para ver cuáles son esos casos y compararlos
    nits_problema = inconsistencias['ID_EMPRESA']
    casos_raros = df_donante[df_donante['ID_EMPRESA'].isin(nits_problema)][['ID_EMPRESA', 'EMPRESA']].drop_duplicates()
    display(casos_raros)
else:
    print("Cada NIT tiene un único nombre de empresa asociado.")
```
